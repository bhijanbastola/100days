```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv("/Users/bhijanbastola/Desktop/intern/Data/train-2.csv")
```

1 . Bar Chart

```python
df.Survived.value_counts().plot(kind="bar",
title="Count of Survived Passenger",
figsize=(5,5),
legend=True,
xlabel="Survived(0 = True, 1 = False)",
ylabel="No.of Passenger",
yticks=range(0,551,50),
color=["Red","Green"],
)
```

```python
ax=df.value_counts(subset=["Sex","Survived"]).plot(kind="bar",
title="Count of Survived Passenger",
figsize=(5,5),
legend=True,
xlabel="Survived(0 = True, 1 = False)",
ylabel="No.of Passenger",
yticks=range(0,551,50),
color=["Red","Green"],
)
ax.bar_label(ax.containers[0],padding=3)
```

```python
pd.crosstab(df.Survived,
            df.Sex).plot(kind="bar",
title="Count of Survived Passenger",
figsize=(5,5),
legend=True,
xlabel="Survived(0 = True, 1 = False)",
ylabel="No.of Passenger",
yticks=range(0,551,50),
color=["Red","Green"],
)
```

2 . Pie Chart

```python
groupby_embarked=df.groupby(by="Embarked")
groupby_embarked.count().plot(
    kind="pie",
    y="Survived",
    title="Count of passenger according to the PClass",
    autopct="%1.1f%%",
    startangle=0,figsize=(4,4)
)
```

```python
pd.crosstab(df.Survived,
            df.Sex).plot(kind="pie",
title="Count of Survived Passenger",
figsize=(8,4),
legend=True,
# xlabel="Survived(0 = True, 1 = False)",
# ylabel="No.of Passenger",
# yticks=range(0,551,50),
color=["Red","Green"],
subplots=True,
autopct="%1.1f%%",
)
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df, x="SalePrice", kde=True, bins=30)
plt.title("Distribution of House Sale Prices")
plt.xlabel("Sale Price ($)")
plt.ylabel("Count")
plt.show()
```

This histogram illustrates the univariate distribution and Kernel Density Estimate (KDE) of **`SalePrice`**, revealing a pronounced positive (right) skewness. The vast majority of properties cluster tightly in the lower-to-mid price range between $100,000 and $200,000, peaking around $135,000–$160,000 before trailing off significantly toward luxury homes reaching up to $750,000. Because this long upper tail causes non-normal target distribution and heteroscedasticity, applying a log transformation like `np.log1p(df['SalePrice'])` is a necessary preprocessing step to stabilize variance and improve regression model performance.

```python
sns.boxplot(x="OverallQual", y="SalePrice", data=df)

plt.title("SalePrice by Overall Quality")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price")
plt.show()
```

This box plot demonstrates a strong positive, non-linear relationship between **`OverallQual`** (rated on an ordinal scale from 1 to 10) and house **`SalePrice`**. As overall quality increases, median sale prices consistently rise—starting around $50,000 for a quality rating of 1–2 and reaching over $400,000 for a rating of 10. Additionally, the variance (interquartile range) and presence of upper-bound outliers expand significantly at higher quality ratings (especially scores 7–10), indicating greater price dispersion for premium properties compared to lower-quality homes.

```python
sns.scatterplot(df,x="GrLivArea",y="SalePrice")
plt.title("SalePrice by GrLivArea Quality")
plt.xlabel("GrLivArea")
plt.ylabel("Sale Price")
plt.show()
```

This scatter plot illustrates a strong positive linear relationship between **`GrLivArea`** (above-grade living area in square feet) and **`SalePrice`**, showing that home values generally scale upward as living space increases. The bulk of observations cluster between 800 sq ft and 2,800 sq ft, maintaining a steady price trend. Notably, the plot reveals **two severe high-leverage outliers** in the bottom-right corner—houses exceeding 4,000 sq ft that sold for unusually low prices below $200,000. These specific anomalies represent partial sales or unusual property conditions, and because their high leverage can heavily skew linear regression models, removing them (`GrLivArea > 4000` & `SalePrice < 200000`) is a crucial preprocessing step in EDA.

```python
plt.figure(figsize=(30,30))
sns.heatmap(df[numerical_columns].corr(),annot=True)
```

 This correlation matrix visualizes the linear relationships among numerical features in the Ames Housing dataset, highlighting the primary drivers of **`SalePrice`**. The strongest positive predictors for target price are **`OverallQual`** ($r = 0.79$), **`GrLivArea`** ($r = 0.71$), **`GarageCars`** ($r = 0.64$), and **`TotalBsmtSF`** ($r = 0.61$), demonstrating that material quality and square footage dominate property valuation. The heatmap also reveals notable multicollinearity between redundant feature pairs, such as `GarageCars` and `GarageArea` ($r = 0.88$) as well as `TotalBsmtSF` and `1stFlrSF` ($r = 0.82$), while temporal/miscellaneous features like `YrSold` and `MiscVal` show near-zero correlation ($\vert{}r\vert{} < 0.05$) with house prices.
